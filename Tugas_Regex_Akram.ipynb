```python
import re
```

```python
teks1 = "Pembayaran untuk pesanan #INV-2026-08912 telah berhasil diverifikasi pada pukul 14.30 WIB."
pola1 = r"INV-\d{4}-\d+"
print("1.", re.findall(pola1, teks1))
```

```python
teks2 = "Total belanjaan Anda adalah Rp 450.000 dengan diskon promo sebesar Rp 50.000."
pola2 = r"Rp\s?\d{1,3}(?:\.\d{3})*"
print("2.", re.findall(pola2, teks2))
```

```python
teks3 = "Transaksi menggunakan kartu kredit 4532-1234-5678-9012 telah disetujui."
pola3 = r"\b(\d{4})-(\d{4})-(\d{4})-(\d{4})\b"
hasil3 = re.sub(pola3, r"XXXX-XXXX-XXXX-\4", teks3)
print("3.", hasil3)
```

```python
teks40 = "brg sdh smp dgn slmt, tq seller"
kamus_singkatan = {
    r"\bbrg\b": "barang",
    r"\bsdh\b": "sudah",
    r"\bsmp\b": "sampai",
    r"\bdgn\b": "dengan",
    r"\bslmt\b": "selamat",
    r"\btq\b": "terima kasih",
}
hasil40 = teks40
for pola, ganti in kamus_singkatan.items():
    hasil40 = re.sub(pola, ganti, hasil40, flags=re.IGNORECASE)
print("40.", hasil40)
```
